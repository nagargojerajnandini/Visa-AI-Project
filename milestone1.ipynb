```python
import pandas as pd
import numpy as np

np.random.seed(42)  # reproducibility
n = 1000
```

```python
education_levels = ["High School", "Bachelors", "Masters", "PhD"]
visa_types = ["Student", "Work", "Tourist"]
countries = ["India", "USA", "UK", "Canada", "Australia", "Germany"]
yes_no = ["Yes", "No"]
```

```python
data = {
    "age": np.random.randint(18, 50, n),
    "annual_income": np.random.randint(5000, 60000, n),
    "education_level": np.random.choice(education_levels, n),
    "work_experience": np.random.randint(0, 15, n),
    "visa_type": np.random.choice(visa_types, n),
    "country": np.random.choice(countries, n),
    "documents_complete": np.random.choice(yes_no, n),
    "interview_score": np.random.randint(1, 10, n),
    "previous_visa_rejection": np.random.choice(yes_no, n),
}
```

```python
df = pd.DataFrame(data)
```

```python
df["visa_status"] = np.where(
    (df["documents_complete"] == "Yes") &
    (df["interview_score"] > 5) &
    (df["previous_visa_rejection"] == "No"),
    "Approved",
    "Rejected"
)
```

```python
df["processing_time_days"] = np.where(
    df["visa_status"] == "Approved",
    np.random.randint(15, 35, n),
    np.random.randint(30, 60, n)
)
```

```python
df.to_csv("visa_dataset_1000.csv", index=False)
```
